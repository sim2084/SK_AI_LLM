# LLM 학습 파이프라인

## 1. LLM 학습 파이프라인 개요

### 1-1. LLM 학습의 정의

대규모 언어 모델(LLM)의 학습은 언어를 이해하고 생성하는 범용 인공지능을 구축하는 핵심 과정이다. 단순히 텍스트를 암기하는 수준을 넘어, 언어의 구조, 의미, 지식, 그리고 사용 맥락을 단계적으로 내재화하는 과정을 거친다.

### 1-2. 전체 학습 흐름

LLM 학습 파이프라인은 크게 세 단계로 구분할 수 있다. 먼저 **사전학습(Pre-training)** 단계에서는 대규모 비라벨 텍스트를 통해 언어의 일반적인 패턴과 세계 지식을 습득한다. 이어지는 **파인튜닝(Fine-tuning & Optimization)** 단계에서는 특정 목적이나 도메인에 맞게 모델을 정교하게 조정하고 효율화한다. 마지막으로 **정렬(Alignment)** 단계에서는 인간의 가치, 의도, 선호에 부합하도록 모델의 응답 행동을 교정하여 실질적인 활용성을 높인다.

> 이러한 흐름은 **범용 언어 능력 습득**에서 시작하여 **목적별 전문화**를 거쳐 **인간 친화적 조정**으로 이어지는 일련의 진화 과정이라고 할 수 있다.

### 1-3. 단계적 학습의 필요성

학습을 단계별로 나누는 이유는 각 단계가 수행하는 역할이 명확하기 때문이다. 사전학습이 “언어를 아는 모델”을 만든다면, 파인튜닝은 이를 “쓸 수 있는 모델”로 탈바꿈시키고, 정렬 단계에서는 비로소 “사람과 함께 안전하게 쓸 수 있는 모델”이 완성된다.

> 이러한 분업 구조를 통해 대규모 모델을 효율적으로 재사용할 수 있으며, 학습 비용을 절감하는 동시에 모델의 안전성과 신뢰성을 체계적으로 확보할 수 있다는 장점이 있다.

## 2. 사전학습(Pre-training) - 언어 지능의 기초

### 2-1. 사전학습의 역할

사전학습은 LLM이 언어의 기본 구조와 일반적인 세계 지식을 학습하는 최초의 관문이다. 이 단계에서는 별도의 라벨이 없는 대규모 텍스트 데이터를 활용하며, 사람이 직접 정답을 제공하지 않는 방식으로 진행한다. 사전학습의 핵심 목적은 모델이 언어를 확률적으로 이해하고 다음 문맥을 예측하는 근본적인 능력을 갖추도록 하는 데 있다.

### 2-2. 학습 원리: 자기지도학습

사전학습은 주로 **자기지도학습(Self-supervised Learning)** 방식을 채택한다. 이는 사람이 명시적으로 정답을 주지 않아도 모델이 텍스트 자체에서 스스로 정답을 찾아 학습하는 방식이다. 문장 내의 단어 관계나 구조를 스스로 파악하며 학습 데이터에 담긴 지식을 흡수한다.

### 2-3. 자기회귀 언어 모델링

사전학습의 가장 핵심적인 태스크는 **다음 토큰 예측(Next Token Prediction, NTP)**이다. 이는 주어진 앞 문맥을 바탕으로 다음에 올 토큰을 확률적으로 예측하는 과정이다.

수식으로는 다음과 같이 표현할 수 있다.

$$
P(w_1, w_2, \cdots, w_n) = \prod_{i=1}^{n} P(w_i \mid w_1, \cdots, w_{i-1})
$$

모델은 문장을 한꺼번에 이해하는 대신, 이전 토큰들을 조건으로 한 확률의 연쇄로 파악한다. 이 과정을 반복하면서 문법적 규칙, 단어의 의미, 그리고 복잡한 문맥의 흐름을 동시에 학습하게 된다.

### 2-4. 손실함수: 교차 엔트로피 손실

> 모델 학습의 목표는 예측의 정확도를 극대화하는 것이다. 이를 위해 모델이 예측한 토큰 분포와 실제 데이터상의 정답 토큰 사이의 차이를 최소화하는 방향으로 학습을 진행한다. 이때 활용하는 지표가 **교차 엔트로피 손실(Cross-Entropy Loss)**이다. 모델의 예측이 정답에 가까워질수록 손실값은 줄어들고, 반대의 경우에는 손실값이 커지면서 모델이 더 정교하게 조정되도록 유도한다.

### 2-5. 사전학습 데이터: 대규모 말뭉치

사전학습에는 방대하고 다양한 텍스트 코퍼스가 사용된다. 주요 데이터 소스로는 웹페이지 데이터를 비롯하여 서적, 학술 논문, 뉴스 기사, 위키피디아 등이 포함된다. 이렇게 폭넓은 데이터를 활용하는 이유는 특정 도메인에 매몰되지 않고 범용적인 언어 지식과 세계에 대한 보편적인 이해를 습득하기 위해서이다.

### 2-6. 기반 아키텍처: 트랜스포머

현대 LLM의 사전학습은 2017년 Google이 제안한 **트랜스포머(Transformer)** 아키텍처를 그 근간으로 한다. 트랜스포머의 핵심은 기존의 순차적인 처리 방식에서 벗어나 **어텐션 메커니즘(Attention Mechanism)**만으로 모델을 구성했다는 점이다.

트랜스포머 아키텍처는 몇 가지 결정적인 장점을 제공한다. 우선 데이터 파이프라인의 병렬화가 가능해져 대규모 데이터를 학습시키는 데 매우 유리하다. 또한 **셀프 어텐션(Self-Attention)** 기술을 통해 문장 내에서 멀리 떨어진 단어 사이의 관계, 즉 **장거리 의존성**을 효과적으로 포착할 수 있다. 마지막으로 모델의 크기와 데이터의 양이 늘어날수록 성능이 비례해서 향상되는 뛰어난 **확장성**을 보여준다.

### 2-7. 사전학습의 결과와 한계

사전학습이 완료된 모델은 언어의 복잡한 구조와 문법을 깊이 있게 이해하며, 학습 데이터에 포함된 방대한 세계 지식을 내재하게 된다. 이를 통해 다양한 도메인과 여러 분야의 작업에 유연하게 적용될 수 있는 범용적인 능력을 갖추게 된다.

하지만 사전학습만으로는 실무에 바로 투입하기에 한계가 있다. 모델이 아직 특정 작업이나 사용자의 구체적인 의도에 정밀하게 맞춰져 있지 않으며, 인간의 가치관이나 안전 기준에 부합하는 응답을 생성하는 능력도 부족하기 때문이다. 이러한 미비점을 보완하여 모델을 더욱 완성도 있게 만들기 위해 다음 단계인 파인튜닝과 정렬 과정이 필요하다.

## 3. 파인튜닝 및 최적화(Fine-tuning & Optimization)

### 3-1. 파인튜닝의 개념과 필요성

> 파인튜닝은 사전학습된 범용 모델을 특정 목적이나 도메인에 맞게 미세 조정하는 필수적인 단계이다. 사전학습이 모델에게 전반적인 언어 지식을 부여한다면, 파인튜닝은 이를 바탕으로 “실질적인 업무를 수행하는 능력”을 길러준다.

사전학습 모델은 방대한 지식을 갖추고 있으나, 특정 태스크의 의도나 형식을 정확히 따르지 못하거나 의료, 법률, 코드와 같은 전문 분야에서는 성능이 제한적일 수 있다. 파인튜닝을 통해 답변의 정확도를 높이고, 출력 스타일을 제어하며, 특정 업무에 최적화된 전문성을 확보할 수 있다.

### 3-2. 파인튜닝의 주요 방식

모델을 조정하는 방식은 크게 전체 파라미터를 학습시키는 방법과 효율성에 집중하는 방법으로 나뉜다.

#### 1. Full Fine-tuning

모델의 모든 파라미터를 업데이트하는 방식이다. 모델을 완전히 재설계하거나 이론적으로 도달할 수 있는 최고의 성능을 목표로 할 때 적합하다. 다만, 모든 가중치를 수정하므로 막대한 컴퓨팅 자원과 메모리가 필요하다는 부담이 있다.

#### 2. PEFT(Parameter-Efficient Fine-Tuning)

> 전체 모델의 가중치는 고정해 둔 채, 일부 핵심 파라미터나 추가적인 모듈만을 학습시키는 방식이다. 메모리를 획기적으로 절약할 수 있고 학습 속도가 빠르며, 여러 태스크에 유연하게 대응할 수 있다는 장점이 있다.

- **LoRA(Low-Rank Adaptation)**: 거대한 가중치 행렬을 저차원 행렬의 결합으로 근사하여 소량의 파라미터만 학습시킨다.
- **QLoRA(Quantized LoRA)**: 4비트 양자화 기술을 LoRA와 결합하여, 대형 모델을 단일 GPU와 같은 제한된 환경에서도 튜닝할 수 있게 한다.
- **Adapter-based Tuning**: 기존 모델 구조를 유지하면서 사이에 작은 어댑터 모듈을 삽입하여 해당 부분만 학습시킨다.

### 3-3. 지식 증류를 통한 모델 효율화

파인튜닝과 병행하여 성능은 유지하면서 모델을 가볍게 만드는 **지식 증류(Knowledge Distillation)** 기법도 널리 쓰인다. 이는 성능이 우수한 대형 모델(Teacher)의 지식을 작은 모델(Student)로 압축하여 전달하는 방식이다. 이를 통해 추론 속도를 높이고 배포 비용을 절감하며, 모바일이나 엣지 디바이스에서도 LLM을 원활히 구동할 수 있다.

### 3-4. 파인튜닝 단계의 성과

> 이 단계를 거친 모델은 특정 태스크에 최적화된 성능과 효율성을 갖추게 된다. 하지만 여전히 인간의 가치 판단이나 안전성, 선호도에 따른 미세한 톤 조정이 부족할 수 있다. 이러한 격차를 메우기 위해 마지막 단계인 **정렬(Alignment)**이 수행된다.

## 4. 정렬(Alignment) - 가치와 선호의 조율

### 4-1. Alignment의 개념과 필요성

정렬(Alignment)은 모델의 행동과 출력이 인간의 가치, 의도, 선호에 완벽히 부응하도록 세밀하게 조정하는 과정이다. 사전학습과 파인튜닝이 “정답을 잘 맞히는 능력”에 집중한다면, 정렬은 “사람이 원하는 방식으로 안전하게 소통하는 능력”에 초점을 맞춘다.

정렬 단계를 거치지 않은 모델은 기술적으로는 정답을 내놓더라도 부적절하거나 편향된 답변을 생성할 위험이 있으며, 사용자의 미세한 요구 사항을 놓치기 쉽다. 따라서 정렬은 모델의 안전성, 신뢰성, 유용성을 최종적으로 확보하는 매우 중요한 과정이다.

### 4-2. 정렬 기법의 진화: RLHF와 그 대안들

#### RLHF(Reinforcement Learning from Human Feedback)

가장 전통적인 정렬 방식은 인간의 피드백을 강화학습에 직접 활용하는 **RLHF**이다.

1. **보상 모델(Reward Model) 학습**: 인간이 모델의 여러 응답을 비교하여 선호도를 표시하면, 이를 바탕으로 모델의 답변이 얼마나 훌륭한지 점수를 매기는 보상 모델을 만든다.
2. **강화학습 수행**: 보상 모델이 주는 점수를 극대화하는 방향으로 LLM을 학습시킨다. 이때 모델의 본래 능력이 훼손되지 않도록 기존 모델에서 크게 벗어나지 않게 관리하는 것이 핵심이다.

RLHF의 핵심 알고리즘으로는 주로 **PPO(Proximal Policy Optimization)**가 사용된다. PPO는 모델의 업데이트가 기존 정책에서 지나치게 멀어지지 않도록 제한함으로써 안정적인 강화학습을 가능하게 한다. 이를 통해 복잡한 인간의 선호를 정교하게 반영하여 강력한 정렬 효과를 거둘 수 있지만, 보상 모델 학습을 포함한 전체 파이프라인의 복잡도가 높고 연산 비용이 많이 든다는 점이 과제로 꼽힌다.

비록 RLHF가 강력한 정렬 효과를 보여주지만, 보상 모델을 학습시키는 비용이 크고 파이프라인이 복잡하다는 단점이 있다.

#### 보상 모델이 필요 없는 최신 기법

최근에는 RLHF의 복잡함을 해결하기 위해 보상 모델 없이 직접 선호를 반영하는 기법들이 주목받고 있다.

> - **DPO(Direct Preference Optimization)**: 별도의 보상 모델 없이 선호 데이터 자체를 직접 최적화에 사용한다. 구조가 단순하고 학습이 안정적이며 구현 난이도가 낮아 널리 채택되고 있다.
> - **ORPO(Odds Ratio Preference Optimization)**: 지도 미세 조정(SFT)과 정렬 단계를 하나로 통합하여 학습 효율을 극대화한다. 확률 비율 기반의 최적화를 통해 데이터와 연산 비용을 줄이면서도 뛰어난 정렬 성능을 보여준다.

### 4-3. 정렬 단계의 완성

정렬이 완료된 모델은 비로소 사용자의 실제 의도를 깊이 있게 이해하고, 안전하며 일관된 응답을 생성하는 능력을 갖추게 된다. 이는 단순히 성능이 좋은 인공지능을 넘어, 실제 서비스 환경에서 사람과 신뢰를 쌓으며 상호작용할 수 있는 준비가 되었음을 의미한다.
