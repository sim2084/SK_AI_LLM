# AI 에이전트란?
**환경과 상호작용하며 목표를 달성하기 위해 스스로 다음 작업을 판단하고 수행하는 소프트웨어**

**핵심 특징**
- ✅ **자율성**: 사람이 목표를 제시하면 필요한 단계를 스스로 판단하여 실행
- ✅ **지능적 판단**: 상황에 따라 적절한 다음 행동을 선택
- ✅ **도구 활용**: 검색, 계산, API, 데이터베이스 등 외부 도구와 시스템을 연동하여 작업 수행
- ✅ **상태 관리**: 이전 대화, 중간 결과, 작업 진행 상황을 바탕으로 다음 단계를 결정

**동작 예시**
> **요청**: "200만원 예산으로 10월 도쿄 여행 일정 짜줘"

**AI 에이전트가 하는 일**:
1. 항공편 후보 검색
2. 숙소 후보 비교
3. 이동 수단과 관광 일정 조사
4. 예산 내에서 가능한 일정 조합 생성
5. 사용자가 승인하면 캘린더 등록 또는 예약 단계로 연결

> 예약, 결제, 캘린더 등록처럼 실제 서비스에 영향을 주는 작업은 외부 API 권한과 사용자 승인이 필요하다.  
> 따라서 실무에서는 중요한 작업 앞에 사용자 확인 단계(Human-in-the-loop)를 두는 것이 안전하다.

**핵심 구성요소**
```
🤖 Agent (에이전트) - 목표를 보고 다음 행동을 결정하는 실행 주체
🧠 Memory - 단기/장기 기억 저장
🛠️ Tools - 외부 도구들 (검색, 계산기, API 등)
📋 Planning - 작업 계획 수립
⚡ Action - 실제 작업 수행
🔁 Feedback - 실행 결과 확인 및 다음 행동 조정
```


---

## 작동 과정
1. 1단계: 목표 분해  
복잡한 목표를 작은 실행 가능한 작업들로 나눔

2. 2단계: 정보 수집  
필요한 데이터를 인터넷, 문서, 데이터베이스, 외부 시스템에서 검색

3. 3단계: 작업 실행  
계획한 작업을 순서대로 수행하고 중간 결과를 저장

4. 4단계: 결과 검증  
결과가 목표에 맞는지 확인하고 필요하면 다시 수행하거나 사용자에게 확인 요청

# LangGraph

### LangGraph 소개
- **LangGraph란?** LangGraph는 LLM 애플리케이션, 에이전트, 워크플로우를 **그래프 구조**로 구성하고 실행 흐름과 상태를 제어할 수 있게 해주는 **상태 기반 그래프 실행 프레임워크**입니다.
- **배경:** 하나의 LLM 호출이나 단순한 체인으로 처리하기 어려운 작업에서는 분기, 반복, 도구 호출, 사용자 승인, 상태 저장 같은 흐름 제어가 필요합니다. LangGraph는 이러한 복잡한 흐름을 노드와 엣지로 명시적으로 설계할 수 있게 해줍니다.
- **주요 목적:** 대화형 에이전트, 복잡한 태스크 자동화, RAG 파이프라인, 사용자 정의 LLM 워크플로우 등을 **확장 가능하고 안정적인 구조**로 구축하는 토대를 제공합니다.

### 철학과 배경 (Why LangGraph?)
- **워크플로우와 에이전트 제어:** LangGraph는 정해진 순서대로 실행되는 워크플로우뿐 아니라, 상황에 따라 다음 행동을 동적으로 결정하는 에이전트 흐름도 구성할 수 있습니다.
- **상태 기계 관점:** LangGraph의 그래프 접근법은 **상태 기계(State Machine)** 개념과 유사합니다. 현재 상태를 기준으로 어떤 노드를 실행할지 결정하고, 각 노드는 실행 결과를 다시 상태에 반영합니다. 이러한 **구조적 설계**를 통해 복잡한 상호작용을 체계적으로 제어할 수 있습니다.
- **멀티 에이전트 활용:** LangGraph는 멀티 에이전트 협업에도 사용할 수 있습니다. 다만 LangGraph가 멀티 에이전트 전용 도구인 것은 아닙니다. 단일 에이전트, 일반 워크플로우, RAG 흐름, Human-in-the-loop 구조에도 사용할 수 있습니다.
- **기존 체인의 한계 보완:** 단순 체인은 보통 정해진 순서대로 실행됩니다. 반면 LangGraph는 **분기, 반복, 조건부 실행, 상태 저장, 실행 재개** 같은 흐름을 표현할 수 있어 복잡한 AI 애플리케이션에 적합합니다.

### 그래프 구조 – 노드(Node)와 엣지(Edge)
- **노드 (Node):** 그래프의 노드는 하나의 **작업 단위**를 나타냅니다. 노드는 단순 Python 함수일 수도 있고, LLM 호출, Tool 실행, 검색 함수, 검증 함수, 또는 하나의 에이전트일 수도 있습니다. 예를 들어, 질문 분류 노드, 문서 검색 노드, 답변 생성 노드, 결과 검증 노드 등이 노드가 됩니다.
- **엣지 (Edge):** 노드 간 **이동 경로**를 나타냅니다. 한 노드의 실행이 끝난 뒤 어느 다음 노드로 이동할지를 결정합니다. 엣지는 **방향성**을 가지며, **조건부 제어**가 가능합니다. 즉, 현재 상태나 메시지 내용에 따라 다른 경로를 선택할 수 있습니다. 이를 통해 분기와 반복을 구현할 수 있습니다.
- **그래프 vs 순차 체인:** 전통적인 체인(chain)은 고정된 순서를 따르지만, **그래프 구조**에서는 유연한 경로 선택과 **분기/반복 흐름**이 가능합니다. 이를 통해 복잡한 워크플로우를 자연스럽게 표현할 수 있습니다.

### 상태 기반 에이전트 설계
- **공유 상태 (State):** LangGraph의 핵심은 그래프 전체가 공유하는 **상태(State)**입니다. 이 상태에는 사용자 입력, 대화 메시지, 검색 결과, 중간 판단 결과, 작업 완료 여부, 최종 답변 후보 등이 포함될 수 있습니다.
- **State와 Memory의 차이:** State는 현재 그래프 실행 중에 노드들이 읽고 쓰는 작업 상태입니다. Memory는 이 상태나 대화 이력을 실행 이후에도 보존하거나, 세션을 넘어 장기적으로 활용하는 개념입니다. 따라서 State를 단순히 “메모리”라고만 이해하면 checkpoint, short-term memory, long-term memory를 배울 때 혼동될 수 있습니다.
- **노드 실행 모델:** 각 노드는 현재 State를 입력으로 받아 자신의 역할에 맞는 작업을 수행하고, State의 일부를 업데이트합니다. 예를 들어, ‘검색’ 노드는 질문을 받아 검색 결과를 State에 추가하고, ‘답변 생성’ 노드는 State에 있는 검색 결과를 읽어 답변을 생성합니다.
- **상태 업데이트와 전이:** 노드가 작업을 마치면 상태를 갱신하고, **다음으로 어느 노드로 이동할지 결정**합니다. LangGraph에서는 이 전이를 명시적으로 정의할 수 있으며, 각 노드의 실행 결과나 State 값에 따라 다음 경로를 바꿀 수 있습니다. 예를 들어, 상태에 `needs_search=True`이면 검색 노드로 이동하고, `needs_search=False`이면 답변 생성 노드로 이동하는 식입니다.

### 메시지 흐름과 제어 흐름
- **메시지 전달:** LangGraph 그래프의 동작 흐름은 **상태를 중심으로 한 메시지 흐름**으로 이해할 수 있습니다. 사용자의 입력 메시지가 그래프에 들어오면, 시작 노드가 이를 처리하고 결과를 State에 반영합니다. 이후 정의된 엣지를 따라 다음 노드가 실행됩니다.
- **제어 흐름:** 엣지에는 **조건부 로직**을 부여할 수 있습니다. 예를 들어, 특정 노드의 출력 내용에 따라 다른 노드로 분기하거나, 그래프 실행을 종료(END)할 수 있습니다. 별도의 “라우터(router)” 노드를 두어 메시지 내용을 해석하고 다음 경로를 결정하기도 합니다.
- **예시 – Router 패턴:** 하나의 라우터 노드가 현재 질문과 State를 보고 “검색 필요”, “도구 호출 필요”, “바로 답변 가능”, “종료” 등의 신호를 결정하면, 각 신호에 대응되는 다음 노드로 이동합니다. 예컨대, **검색 필요**이면 검색 노드로, **도구 호출 필요**이면 Tool 노드로, **바로 답변 가능**이면 답변 생성 노드로, **종료**이면 END로 가는 식입니다.
- **예시 – 검증 루프:** 답변 생성 후 검증 노드가 결과를 확인하고, 근거가 부족하면 다시 검색 노드로 되돌릴 수 있습니다. 반대로 충분하면 최종 답변으로 종료할 수 있습니다. 이런 반복 구조는 일반적인 순차 체인보다 그래프 구조에서 더 명확하게 표현됩니다.

### 실무 적용 고려사항
- **종료 조건 설정:** 그래프 흐름은 잘못 설계하면 무한 루프에 빠질 수 있습니다. **그래프 종료 조건(END)**을 명확히 정의해야 합니다. 예를 들어, 답변 검증이 통과되면 종료하고, 일정 횟수 이상 재시도하면 사용자에게 확인을 요청하는 규칙을 둘 수 있습니다.
- **프롬프트 설계:** LLM을 사용하는 노드에는 **명확한 역할 지시**를 하는 프롬프트가 필요합니다. 각 노드가 자신의 책임만 수행하도록 지침을 줘야 합니다. 예를 들어, 한 노드는 “질문을 분석하여 검색 필요 여부 판단”, 다른 노드는 “검색 결과를 바탕으로 답변 생성” 식으로 역할을 분리할 수 있습니다.
- **오버헤드와 비용:** 여러 노드가 순차 또는 반복적으로 LLM을 호출하면 **API 호출 비용**과 **응답 지연**이 늘어날 수 있습니다. 실무 적용 시에는 필요한 최소한의 노드로 구성하고, 반복 횟수 제한이나 타임아웃 같은 제어를 넣어 **효율성**을 관리해야 합니다.
- **디버깅과 모니터링:** 그래프 구조상 흐름이 복잡해질 수 있으므로, **중간 상태 로깅**이나 **실행 경로 추적**이 중요합니다. LangGraph로 실행되는 동안 어떤 경로를 거쳤는지, 각 노드의 입력/출력은 무엇이었는지 추적하면 디버깅에 도움이 됩니다. LangSmith, Langfuse 같은 관측 도구를 함께 활용할 수 있습니다.
- **Checkpoint와 실행 재개:** LangGraph는 실행 상태를 저장하는 checkpoint 구조를 활용할 수 있습니다. 이를 통해 대화 이력을 이어가거나, 중간에 멈춘 작업을 나중에 이어서 실행하는 구조를 만들 수 있습니다.
- **Human-in-the-loop:** 예약, 결제, 삭제, 배포처럼 중요한 작업은 AI가 바로 실행하지 않고 사용자 승인 단계를 거치게 설계하는 것이 좋습니다. LangGraph에서는 중간 상태를 확인하고 사람이 개입한 뒤 다시 흐름을 이어가는 구조를 만들 수 있습니다.
- **모델 한계 및 검증:** 각 노드는 LLM의 한계를 지닙니다. 검색 결과를 잘못 해석하거나, 도구 결과를 부정확하게 요약하거나, 잘못된 다음 경로를 선택할 수 있습니다. 따라서 중요한 작업에는 검증 노드, 재시도 조건, 사용자 확인 단계를 포함하는 것이 좋습니다.

### 활용 사례 및 마무리
- **QA 시스템:** 사용자의 질문을 받아 **질문 분석 노드**가 검색 필요 여부를 판단하고, **검색 노드**가 문서를 찾고, **답변 생성 노드**가 결과를 요약하여 답변하는 시스템을 만들 수 있습니다. 필요하면 **검증 노드**를 추가하여 근거가 부족한 답변을 다시 검색하도록 만들 수도 있습니다.
- **코드 생성 협업:** **코드 작성 노드**와 **코드 검토 노드**가 번갈아 작동하여 고품질 코드를 산출합니다. 작성자가 코드를 생성하면 검토자가 오류를 지적하고 수정 요청을 하며, 다시 작성자가 개선하는 **반복 루프**를 그래프를 통해 구현할 수 있습니다.
- **문서 요약 파이프라인:** 긴 문서를 **분할 요약 노드**가 부분 요약하고, **통합 요약 노드**가 이를 종합하여 최종 요약을 생성합니다. 이처럼 요약 작업을 단계별 노드로 분리하면 긴 입력도 구조적으로 처리할 수 있습니다.
- **업무 자동화 에이전트:** 대화형 비서에서 질의 유형에 따라 일정 관리, 이메일 작성, 문서 검색, 데이터 분석 같은 노드로 전환하며 처리할 수 있습니다. 중요한 실행 단계 앞에는 사용자 승인 노드를 넣어 안전성을 높일 수 있습니다.
- **멀티 에이전트 협업:** 계획 수립 에이전트가 전체 작업을 나누고, 검색 에이전트, 작성 에이전트, 검토 에이전트가 각각 자신의 역할을 수행하는 구조를 만들 수 있습니다. LangGraph의 그래프 접근법은 이러한 확장적인 시나리오를 효과적으로 지원합니다.
- **마무리:** LangGraph를 활용하면 AI 워크플로우와 에이전트 시스템을 **구조적이고 안정적으로 설계**할 수 있습니다. 핵심은 “여러 에이전트를 연결한다”가 아니라, **State를 중심으로 노드와 엣지를 설계하여 실행 흐름을 명확하게 제어한다**는 점입니다.
